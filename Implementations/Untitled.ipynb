```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
import joblib
from sklearn.naive_bayes import ComplementNB
```

```python
# Load dataset
data = pd.read_csv('spam_cleaned.csv')
data.columns = ['label', 'message']
data['label'] = data['label'].map({'ham': 0, 'spam': 1})  # Convert labels to binary
```

```python
# Split data
X_train, X_test, y_train, y_test = train_test_split(data['message'], data['label'], test_size=0.2, random_state=42)

```

```python
X_train, X_test, y_train, y_test = train_test_split(
    data['message'], data['label'], test_size=0.2, random_state=42
)
```

```python
# Train a spam detection model
model = make_pipeline(TfidfVectorizer(), MultinomialNB())
model.fit(X_train, y_train)
```

```python
# Save the model
joblib.dump(model, 'spam_model.pkl')
```

```python
# Separate ham and spam
spam = data[data['label'] == 1]  # All spam messages
ham = data[data['label'] == 0].sample(n=len(spam), random_state=42)  # Downsample ham
```

```python
# Merge them back
balanced_data = pd.concat([spam, ham])
```

```python
# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    balanced_data['message'], balanced_data['label'], test_size=0.2, random_state=42
)
```
